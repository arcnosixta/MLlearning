## Tізімдер (Lists)

**Тізімдер** - элементтердің реттелген тізбектеріне арналған контейнер түрі. Тізімдерді бос етіп инициализациялауға болады.

```python
my_list = []
```

немесе кейбір бастапқы элементтермен

```python
my_list = [1, 2, 3]
```

Тізімдердің динамикалық өлшемі бар және оларға ***append*** арқылы қосуға болады

```python
my_list.append(4)
my_list
```

Біз тізімнің жеке элементтеріне де қол жеткізе аламыз (индекстеу 0-ден басталады)

```python
my_list[2]
```

Тізімнің «бөліктеріне» my_list[i:j] командасын пайдаланып та қол жеткізе аламыз, мұндағы i – бөліктің басы (тағы да индекстеу 0-ден басталады) және `j – бөліктің соңы. Мысалы:

```python
my_list[1:3]
```

Екінші индексті өткізіп жіберу бөлік тізімнің соңына дейін жұмыс істеуі керек дегенді білдіреді

```python
my_list[1:]
```

Элементтің тізімде бар-жоғын `in` арқылы тексере аламыз

```python
5 in my_list
```

Тізімнің ұзындығын `len` функциясын пайдаланып алуға болады

```python
len(my_list)
```

## Шартты белгілер

Атауынан көрініп тұрғандай, шартты белгілер шарттың True немесе False болуына байланысты кодты орындау тәсілі болып табылады. Басқа тілдердегідей, Python тілі `if` және `else` тілдерін қолдайды, бірақ `else if` тілі `elif` тіліне қысқартылған, мұны төмендегі мысалдан көруге болады.

```python
my_variable = 5
if my_variable < 0:
  print("negative")
elif my_variable == 0:
  print("null")
else: # my_variable > 0
  print("positive")
```

Мұндағы `<` және `>` – қатаң `кіші` және `үлкен` операторлары, ал `==` – теңдік операторы (айнымалыны тағайындау операторы `=`-мен шатастырмау керек). `<=` және `>=` операторларын `кіші` немесе `тең` салыстырулары үшін пайдалануға болады.

Басқа тілдерден айырмашылығы, код блоктары шегініс арқылы бөлінеді. Мұнда біз 2 бос орын шегінісін қолданамыз, бірақ көптеген бағдарламашылар 4 бос орын шегінісін де пайдаланады. Кодыңызда біркелкілік сақталған жағдайда кез келгені жарайды.

## Циклдер (Loops)

Циклдер – код блогын бірнеше рет орындау тәсілі. Циклдердің екі негізгі түрі бар: while және for циклдері.

While циклі

```python
i = 0
while i < len(my_list):
  print(my_list[i])
  i += 1 # equivalent to i = i + 1
```

For циклі

```python
for i in range(len(my_list)):
  print(my_list[i])
```

Егер мақсат тізім бойынша итерация жасау болса, біз мұны келесідей тікелей жасай аламыз

```python
for element in my_list:
  print(element)
```

## Функциялар (Functions)

Кодты оқу мүмкіндігін жақсарту үшін кодты нақты әрекеттерді орындауға жауапты әртүрлі блоктарға бөлу функциялар арқылы орындалады.

Функция кейбір кірістерді қабылдайды және кейбір шығыстарды қайтару үшін оларды өңдейді.

```python
def square(x):
  return x ** 2

def multiply(a, b):
  return a * b

# Functions can be composed.
square(multiply(3, 2))
```

Кодтың оқылуын жақсарту үшін кейде аргументтерді нақты атау пайдалы

**Жаттығу 1.** Шартты пайдаланып, [relu](https://en.wikipedia.org/wiki/Rectifier_(neural_networks)) функциясын жазыңыз

$\text{relu}(x) = \left\{
   \begin{array}{rl}
     x, & \text{if }  x \ge 0 \\
     0, & \text{otherwise }.
   \end{array}\right.$

```python
def relu(x):
  if x >= 0:
    return x
  else:
    return 0


relu(-3)
```

**Жаттығу 2.** for циклін пайдаланып, тізімдегі сандардың квадраттарының қосындысын есептейтін функция жазыңыз.

---



---



```python
def sum_of_squares(numbers):
   a=0
   for i in range(len(numbers)):
      a=a+numbers[i]**2

   return a



my_numbers = [5, 5, 5, 5]
sum_of_squares(my_numbers)
```

**Жаттығу 3.** For циклін және шартты амалды пайдаланып, вектордағы ең үлкен мәнді қайтаратын функция жазыңыз.

```python
def find_max(vector):
    max_val = vector[0]
    for num in vector:
        if num  max_val:
            max_val = num
    return max_val

my_vector = [3, 77, 2, 9, 5]
print(find_max(my_vector))
```
